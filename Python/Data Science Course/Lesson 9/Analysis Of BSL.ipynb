```python
import matplotlib.pyplot as plt
```

```python
blood_sugar_men = [113, 85, 90, 150, 149, 88, 93, 115, 135, 80, 77, 82, 129]
blood_sugar_women = [67, 98, 89, 120, 133, 150, 84, 69, 89, 79, 120, 112, 100 ]
blood_sugar_kids = [90, 100, 110, 120, 130, 140, 150, 160, 170, 180, 90, 110, 133]

type = [blood_sugar_men, blood_sugar_women, blood_sugar_kids]
colors = ['g', 'r', 'b']
label = ['men', 'women', 'kids']
bins = [80, 100, 125, 150]
plt.xlabel("Blood Sugar Range")
plt.ylabel("total no. of patients")
```

```python
plt.hist(type, bins=bins, rwidth=0.95, color=colors,
         label=label, orientation="horizontal")

plt.title("Blood Sugar Level Chart")
plt.legend()
plt.show()
```
