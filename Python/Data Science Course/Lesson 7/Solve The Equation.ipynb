```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(1, 11, 1)

y1 = (2*x) + 1
y2 = (2*x**2) + 2

plt.figure(figsize=(7, 5))

plt.plot(x, y1, 'g', linewidth=3, label='y=2x+1')
plt.plot(x, y2, 'r', linewidth=3, label='y=2x^2+2')

plt.legend(loc='upper left')
plt.show()
```

```python
x = np.arange(1, 11, 1)

y1 = (2*x) + 1
y2 = (2*x**2) + 2

plt.fill_between(x,y1,y2, color='green', alpha=0.5)
plt.show()
```
