 PART 1: WHT is Seaborn - Import and Load Data

```python
df = sns.load_dataset('penguins')
df = df.dropna()

print("First 5 rows")
print(df.head())
print()
print(df.info())
print()
print(df.describe())
print()
print("Species:", df['species'].unique())
print("Islands:", df['islands'].unique())
```

PART 2: HISTOGRAM

```python
sns.histplot(data=df, x='bpdy_mass_g', bins=20, color='steelblue')
plt.title('Distribution of Penguin Body Mas')
plt.xlabel('Body Mass (grams)')
plt.ylabel('Count')
plt.show()
```

PART 3: KDE PLOT AND FITTING DISTRIBUTION

```python
sns.kdeplot(data=df, x= 'flipper_length_mm', hue='species', fill=True)
plt.title('Flipper Length shape by Species (KDE)')
plt.xlabel('Flipper Length (mm)')
plt.show()

```

PART 4: SCATTER PLOT 


```python
sns.scatterplot(data=df, x='flipper_length_mm', y='body_mass_g', hue='species')
plt.title('Flipper Length vs Body Mass by Species')
plt.xlabel('Flipper Length (mm)')
plt.ylabel('Body Mass (grams)')
plt.show()
```

PART 5 : HEATMAP

```python
corr = df.corr(numeric_only=True)
print("Correlation table:")
print(corr)
print()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap - Penguin Measurements')
plt.show()
```
